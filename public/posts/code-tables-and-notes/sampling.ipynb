## Sampling the estimator

The variance we expect is $\sigma^2/n$; the point of the cell below is
to see whether the simulation agrees.


```python
import numpy as np

rng = np.random.default_rng(20260904)
draws = rng.normal(size=(6, 2000))
means = draws.mean(axis=1)
print(f"observed sd: {means.std(ddof=1):.5f}")
print(f"expected sd: {1 / np.sqrt(2000):.5f}")
```

```python
counts, _ = np.histogram(means, bins=6)
counts
```

```python
import matplotlib.pyplot as plt

plt.bar(range(len(counts)), counts)
plt.show()
```
